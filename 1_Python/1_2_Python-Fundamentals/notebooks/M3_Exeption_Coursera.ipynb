```python
#Exercise 1: Handling ZeroDivisionError
#Type your code here
def safe_divide(numerator,denominator):
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")
        return None
# -----------------------------Test case
#if __name__ == "__main__":
try:
    #numerator = int(input("Enter the numerator value: "))
    #denominator = int(input("Enter the denominator value: "))
    numerator = 1
    denominator = 2 
    print(safe_divide(numerator, denominator))

    numerator = 1
    denominator = 0
    print(safe_divide(numerator, denominator))
except ValueError:
    print("Error: Please enter valid integers.")
```

```python
#Exercise 2: Handling ValueError
import math
def f_square_root(vNum):
    res = None
    try:
        res = math.sqrt(vNum)        
    except ValueError:
        res = "Error: Invalid input! Please enter a positive integer or a float value."
    finally:
        return res
# --------------------------------Test case
#num1=float(input("Enter the number:-"))
num1 = 11
print(f_square_root(num1))
num1 = - 11
print(f_square_root(num1))

#if num1 > 0:
#print(f_square_root(num1))
#else:
#print(""Error: Invalid input! Please enter a positive integer or a float value."")
```

```python
#Exercise 3: Handling Generic Exceptions
def complex_calculation(num):
    result = None
    try:
        result = num / (num - 5)        
    except Exception as e:
        result = "An error occurred during calculation."
    finally:
        return result
# Test case
#user_input = float(input("Enter a number: "))
user_input  = 10
print(f"Result: {complex_calculation(user_input)}")
user_input  = 5
print(f"Result: {complex_calculation(user_input)}")
```
