```python
#Exercise 1
#Extract the required GDP data from the given URL using Web Scraping.
import requests
import pandas as pd


url = "https://web.archive.org/web/20230902185326/https://en.wikipedia.org/wiki/List_of_countries_by_GDP_%28nominal%29"
#url = 'https://en.wikipedia.org/wiki/List_of_largest_banks'
#url = 'https://en.wikipedia.org/wiki/List_of_countries_by_GDP_%28nominal%29'
#local_file = "../output/Largest_economies.csv"

response = requests.get(url)

if response.status_code == 200:
    #with open(local_file, "wb") as f:
    #    f.write(response.content)
    tables = pd.read_html(response.text)
else:
    raise Exception(f"Download failed: {response.status_code}")

#df = pd.read_csv(local_file, header=None)
#print('len(tables) -> ',len(tables))
df1 = pd.DataFrame(tables[1])
#print(df1.head())
df2 = tables[2] # -> with Estimate 
#print('tables[2].head()-----------\n',tables[2].head())
df3 = tables[3]
df2.columns
#df2 = df2.sort_values(
#    by=[
#        ('IMF[1][13]', 'Estimate'),
#        ('World Bank[14]', 'Estimate'),
#        ('United Nations[15]', 'Estimate')
#    ],
#    ascending=[True, True]
#)
df2
```

```python
#rename Column
df2.columns = range(df2.shape[1]) #['Col1','Col2'] # or to int [1,2] or range(df3.shape[1])
df2.columns
```

```python
# Retain columns with index 0 and 2 (name of country and value of GDP quoted by IMF)
df2 = df2[[0,2]]
df2
```

```python
# Retain the Rows with index 1 to 10, indicating the top 10 economies of the world.
df2 = df2.iloc[1:11,:]
df2
```

```python
# Assign column names as "Country" and "GDP (Million USD)"
df2.columns = ['Country','GDP (Million USD)']
df2
```

####    Exercise 2

```python
# Change the data type of the 'GDP (Million USD)' column to integer. Use astype() method.
#import numpy as np
df2['GDP (Million USD)'] = df2['GDP (Million USD)'].astype(int)
#df2 = df2.loc[1,2]
#df2
```

```python
# Convert the GDP value in Million USD to Billion USD
df2[['GDP (Million USD)']] = df2[['GDP (Million USD)']]/1000
df2
```

```python
# Use numpy.round() method to round the value to 2 decimal places.
import numpy as np
df2[['GDP (Million USD)']] = np.round(df2[['GDP (Million USD)']], 2)
df2
```

```python
# Rename the column header from 'GDP (Million USD)' to 'GDP (Billion USD)'
df2.rename(columns = {'GDP (Million USD)' : 'GDP (Billion USD)'})
```
