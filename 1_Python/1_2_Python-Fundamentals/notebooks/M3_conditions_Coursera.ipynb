```python
#Analyze a dataset containing information about renowned athletes, including their names, sports, and notable achievements, to apply conditional logic and filtering techniques
#Write a Python program to check if a player Lionel Messi has more than 10 achievements. If the condition is true, print the player's name, sport, and achievements else print does not have more than 10 achievements.
# Creat dataSet
dSet = [
            {
                "player_name": "Serena Williams",
                "Sport": "Tennis",
                "Achievements": 23
            },
            {
                "player_name": "Lionel Messi",
                "Sport": "Soccer",
                "Achievements": 7
            },
            {
                "player_name": "Michael Phelps",
                "Sport": "Swimming",
                "Achievements": 23
            },
            {
                "player_name": "Usain Bolt",
                "Sport": "Athletics",
                "Achievements": 8
            },
            {
                "player_name": "Roger Federer",
                "Sport": "Tennis",
                "Achievements": 20
            },
            {
                "player_name": "Cristiano Ronaldo",
                "Sport": "Soccer",
                "Achievements": 5
                
            }
]

print(dSet)
#print(dSet[2]['player_name'])
#print(dSet[2]['Sport'])
#print(dSet[2]['Achievements'],type(dSet[2]['Achievements']))
#print(len(dSet), range(len(dSet)))

```

```python
#iRow = 0
dSet.sort(key=lambda row: row['Achievements'])

for iRow in range(0,len(dSet),1):
    #print(iRow,type(i))
    #print(dSet[iRow]['Achievements'])
    if int(dSet[iRow]['Achievements']) > 10:
        print(f"{dSet[iRow]['player_name']} \
              plays {dSet[iRow]['Sport']} \
                and has {dSet[iRow]['Achievements']} achievements.")
    else:
        print(f"{dSet[iRow]['player_name']} does not have more than 10 achievements.")
```

```python
#Write a Python program to check if a player belongs to the sport Tennis or has exactly 20 achievements. If the condition is true, print a success message
for iRow in range(0,len(dSet),1):    
    if int(dSet[iRow]['Achievements']) == 20:
        print(f"{dSet[iRow]['player_name']} \
              plays {dSet[iRow]['Sport']} \
                and has {dSet[iRow]['Achievements']} achievements.")
    #else:
    #    print(f"{dSet[iRow]['player_name']} does not have more than 10 achievements.")
```

```python
#Write a Python program to check if a player has less than 10 achievements and does not play Soccer. Print their details if they meet the criteria.
for iRow in range(0,len(dSet),1):    
    if int(dSet[iRow]['Achievements']) < 10 and dSet[iRow]['Sport'] != 'Soccer' :
        print(f"{dSet[iRow]['player_name']} \
              plays {dSet[iRow]['Sport']} \
                and has {dSet[iRow]['Achievements']} achievements.")
```
