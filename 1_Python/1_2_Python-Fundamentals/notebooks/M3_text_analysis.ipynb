```python
givenstring="Lorem ipsum dolor! diam amet, consetetur Lorem magna. sed diam nonumy eirmod tempor. diam et labore? et diam magna. et diam amet."
# Let's create a class called TextAnalyzer to analyze text.
class TextAnalyzer(object):
    # The __init__ method initializes the class with a 'text' parameter.
    # You will store the provided 'text' as an instance variable.
    def __init__(self, text:str):
        self.sTextAnalyzer = text

    #Step 3: Implement a code to format the text in lowercase
    def fLowerCase(self)-> str:
        return self.sTextAnalyzer.lower()
    #Step 4: Implement a code to count the frequency of all unique words
    def ffreqAll(self,sFreqText = str("")):
        myDictFreq = {}
        if sFreqText == "" : 
            MyDublList = self.sTextAnalyzer.split()
        else:MyDublList = sFreqText.split()
        for key_el in MyDublList:
           myDictFreq[key_el] = MyDublList.count(key_el)
        return myDictFreq
# Test 
obj = TextAnalyzer(givenstring)
print(obj.sTextAnalyzer)
print("Step 3: ",obj.fLowerCase())
print("Step 4: ",obj.ffreqAll(""))
#Step 4.1+: Call the function that counts the frequency of a specific word
sMyTestStr = str("lorem")
print(f"Step 4.1+: word {sMyTestStr} is ", obj.ffreqAll(sMyTestStr))
```
