```python
#Define a dictionary 'x'

x = {'Name': ['Rose','John', 'Jane', 'Mary'], 'ID': [1, 2, 3, 4], 'Department': ['Architect Group', 'Software Group', 'Design Team', 'Infrastructure'], 
      'Salary':[100000, 80000, 50000, 60000]}

#casting the dictionary to a DataFrame
df = pd.DataFrame(x)

#display the result df
df
```

```python
#check the type of x
type(x)
```

```python
#Retrieving the Department, Salary and ID columns and assigning it to a variable z

z = df[['Department','Salary','ID']]
z
```

```python
#Exercise 2: loc() and iloc() functions
#loc[row_label, column_label]
#iloc[row_index, column_index]

# Access the value on the first row and the first column
print(df.iloc[0, 0])
# Access the value on the first row and the third column
print(df.iloc[0,2])
# Access the column using the name
print(df.loc[0, 'Salary'])
```

```python
df2=df
df2=df2.set_index("Name")
#To display the first 5 rows of new dataframe
print(df2.head())
#Now, let us access the column using the name
print("-------------------------- \n ['Jane', 'Salary'] ->",df2.loc['Jane', 'Salary'])
```

```python
#Use the loc() function,to get the Department of Jane in the newly created dataframe df2.
print("['Jane', 'Department'] ->",df2.loc['Jane', 'Department'])
```

```python
#Use the iloc() function to get the Salary of Mary in the newly created dataframe df2.
print("['Mary', 'Salary'] ->",df2.iloc[3, 2])
```

```python
#Exercise 3: Slicing
# let us do the slicing using old dataframe df
print(df)
print(df.iloc[0:2, 0:3])
```

```python
#let us do the slicing using loc() function on old dataframe df where index column is having labels as 0,1,2
print(df)
print(10* '--')
print(df.loc[0:2,'ID':'Department'])
```

```python
#let us do the slicing using loc() function on new dataframe df2 where index column is Name having labels: Rose, John and Jane
df2.loc['Rose':'Jane', 'ID':'Department']
```
