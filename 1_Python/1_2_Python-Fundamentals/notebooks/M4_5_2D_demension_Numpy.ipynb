```python
import numpy as np
a = [[1,2,3,4],[21,22,23,24],[31,32,33,34]]
A = np.array(a)
A #.shape(3,3)
#arr[row][col]
#arr[row][col:bis offset]
```

```python
# Show the numpy array dimensions
print('dimensions of array: ',A.ndim)

# Show the numpy array shape ROWS x COLUMNS 
print('numpy array shape',A.shape) 

# Show the numpy array size
A.size
```

```python
print(A)
# Access the element on the second row and third column
print('\n element A[1, 2] ->',A[1, 2])
# Access the element on the second row and third column
print('A[1][2] ->',A[1][2])
print('rows A[1] ->',A[1])
print('columns A[:][1] ->',A[0:3][1])
# Access the element on the first row and first column
print('A[0][0]->',A[0][0])
# Access the element on the first row and first and second columns
print('\nA[0][0:2]->',A[0][0:2])
# Access the element on the first and second rows and third column
print('\nA[0:2, 2 ->',A[0:2, 2])

print(A[:, 1])       # Accessing a column (2nd column)
```

```python
# Create a numpy array X
X = np.array([[1, 0], [0, 1]]) 
# Create a numpy array Y
Y = np.array([[2, 1], [1, 2]]) 
print(X,'\n-------\n',Y)
# Add X and Y
Z = X + Y
print('\nZ = X + Y->\n',Z)
```

```python
# Create a numpy array Y
Y = np.array([[2, 1], [1, 2]]) 
print(Y)
# Multiply Y with 2
Z = 2 * Y
print('\n2 * Y\n',Z)
```

```python
# Create a numpy array Y
Y = np.array([[2, 1], [1, 2]]) 
print(Y)
# Create a numpy array X
X = np.array([[1, 0], [0, 1]]) 
print(X)
# Multiply X with Y
Z = X * Y
print('Z = X * Y->\n',Z)
```

```python
# Create a matrix A
A = np.array([[0, 1, 1], [1, 0, 1]])
print(A)
# Create a matrix B
B = np.array([[1, 1], [1, 1], [-1, 1]])
print(B)
# Calculate the dot product (умножает и суммирует)
Z = np.dot(A,B)
print('Z = np.dot(A,B)\n',Z)
# Calculate the sine of Z
print('\n np.sin(Z)\n',np.sin(Z))
```
