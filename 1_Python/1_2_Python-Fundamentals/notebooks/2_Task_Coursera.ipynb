```python
# Casting 1.1 to integer will result in loss of information
fVal = 1.1
print(type(fVal), " is ",fVal )
# But
print(type(fVal)," is ",int(fVal) )
```

```python
# Convert a string into an integer
sVal = str("1")
print("Type is ",type(sVal))
print("sVal -> int: ", type((int(sVal))))
```

```python
# Convert a string into an integer with error
sVal = "1 or 2 people"
print("Convert with Err ", int('1 or 2 people'))
```

```python
# Convert the string "1.2" into a float
sVal = "1.2"
print("type is",type(sVal))
print("type is", type(float(sVal)))
```

```python
# Convert an integer to a string
iVal = 1
print("type is",type(iVal),iVal)
print("type is", type(str(iVal)),iVal)
```

```python
# Convert a float to a string
fVal = 1.2
print("type is",type(fVal),fVal)
print("type is", type(str(fVal)),fVal)
```

```python
# Boolean data type
my_val = True
print("True is ",type(my_val), my_val,"is",int(my_val))
my_val = False
print("False is ",type(my_val), my_val,"is",int(my_val))
print("1 is ", bool(1))
print("0 is ", bool(0))
```
