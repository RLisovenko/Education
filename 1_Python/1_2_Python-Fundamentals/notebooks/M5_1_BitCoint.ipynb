```python
#!pip install pycoingecko

from pycoingecko import CoinGeckoAPI

cg = CoinGeckoAPI()

bitcoin_data = cg.get_coin_market_chart_by_id(
    id='bitcoin',
    vs_currency='usd',
    days=30
)

bitcoin_data
```

```python
import pandas as pd
data = pd.DataFrame(bitcoin_data['prices'], columns=['TimeStamp','Price'])
data
```

```python
data['Date'] = pd.to_datetime(data['TimeStamp'], unit = 'ms')
data
```

```python
candlstick_data = data.groupby(data.Date.dt.date).agg({'Price':['min','max','first','last']})
candlstick_data
```

```python
import plotly.graph_objects as go
import plotly.offline as pyo

fig = go.Figure(
    data=[
        go.Candlestick(
            x=candlstick_data.index,
            open=candlstick_data['Price']['first'],
            high=candlstick_data['Price']['max'],
            low=candlstick_data['Price']['min'],
            close=candlstick_data['Price']['last']
        )
    ]
)

fig.update_layout(
    xaxis_rangeslider_visible=False,
    xaxis_title='Date',
    yaxis_title='Price (USD $)',
    title='Bitcoin Candlestick Chart Over Past 30 Days'
)

pyo.plot(
    fig,
    filename='../output/bitcoin_candlestick_graph.html',
    auto_open=False
)
```
