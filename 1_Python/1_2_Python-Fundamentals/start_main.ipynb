# Python Fundamentals — Laboratory Menu

**Completed by:** R. Lisovenko

Open a notebook below and run its **Notebook setup** cell first.
Proceed through the modules in order. This menu does not execute laboratories.
Some exercises require internet access or additional packages. Original exercise
variants and saved results are retained; this is not a guarantee that all external
services or every original code cell will run unchanged.

## Directory Structure

```text
1_2_Python-Fundamentals/
├── notebooks/        # Jupyter notebooks and laboratory exercises
├── data/             # Input datasets and downloaded source files
├── images/           # Images used in exercises
├── output/           # Files and results generated by exercises
├── docs/             # Documentation and certificates
├── start_main.ipynb  # Laboratory menu
├── .gitignore
└── README.md
```

Paths in laboratory code use `../data/`, `../images/`, and `../output/`.
Download exercises refresh their local source files; writing exercises save their
results in `output/`. Existing result files are preserved during organization.
