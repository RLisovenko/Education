```python
import yfinance as yf
import pandas as pd
```

```python
#Using the yfinance Library to Extract Stock Data
#Using the Ticker module we can create an object that will allow us to access functions
#  to extract data. To do this we need to provide the ticker symbol for the stock, 
# here the company is Apple and the ticker symbol is AAPL.
apple = yf.Ticker("AAPL")
```

```python
#wget https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/data/apple.json
import urllib.request

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/data/apple.json"
urllib.request.urlretrieve(url, "apple.json")
```

```python
#Using the attribute info we can extract information about the stock as a Python dictionary.
#apple.json → читается с диска
#json.load → превращает JSON в Python dict
#apple_info → словарь с данными по акции Apple
#последняя строка выводит словарь (в Jupyter)

import json
with open('apple.json') as json_file:
    apple_info = json.load(json_file)
    # Print the type of data variable    
    #print("Type:", type(apple_info))
apple_info
```

```python
#get the 'country' using the key country
apple_info['country']
```

```python
#Extracting Share Price
#A share is the single smallest part of a company's stock that you can buy, 
# the prices of these shares fluctuate over time. Using the history() method 
# we can get the share price of the stock over a certain period of time. 
# Using the period parameter we can set how far back from the present to get data. 
# The options for period are 1 day (1d), 5d, 1 month (1mo) , 3mo, 6mo, 1 year (1y), 2y, 5y, 10y, ytd, and max.
apple_share_price_data = apple.history(period="max")
#apple_share_price_data
apple_share_price_data.head()
```

```python
apple_share_price_data.reset_index(inplace=True)
```

```python
#We can plot the Open price against the Date:
apple_share_price_data.plot(x="Date", y="Open")
```

```python
#Extracting Dividends 
#Using the variable dividends we can get a dataframe of the data. 
# The period of the data is given by the period defined in the 'history` function.

#apple_share_price_data.head()
apple.dividends
```

```python
apple.dividends.plot()
```
