```python
#Problem 3 List all case numbers for crimes involving minors?(children are not considered minors for the purposes of crime analysis)
#%sql  SELECT CASE_NUMBER \
#      FROM CHICAGO_CRIME_DATA \
#      WHERE DESCRIPTION LIKE '%MINOR%'

sql_query = """
SELECT CASE_NUMBER,DESCRIPTION
FROM CHICAGO_CRIME_DATA
WHERE DESCRIPTION LIKE '%MINOR%'
"""
pd.read_sql_query(sql_query, con)
```

```python
#условие	значение
#PRIMARY_TYPE = 'KIDNAPPING'	преступление — похищение
#DESCRIPTION LIKE '%CHILD%'	в описании упоминается ребёнок
sql_query = """
SELECT CASE_NUMBER,PRIMARY_TYPE	,DESCRIPTION
FROM CHICAGO_CRIME_DATA
WHERE PRIMARY_TYPE = 'KIDNAPPING' and DESCRIPTION LIKE '%CHILD%'
"""
pd.read_sql_query(sql_query, con)
```

```python
#Problem 5 List the kind of crimes that were recorded at schools. (No repetitions)
#PRIMARY_TYPE	вид преступления
#DISTINCT	убрать повторения
#LOCATION_DESCRIPTION	где произошло
#LIKE '%SCHOOL%'	ищем слово school
sql_query = """ SELECT DISTINCT CASE_NUMBER,DATE,PRIMARY_TYPE,DESCRIPTION,LOCATION_DESCRIPTION
                FROM CHICAGO_CRIME_DATA
                where LOCATION_DESCRIPTION like '%school%' limit 5
            """
pd.read_sql_query(sql_query, con)
```

```python
import pandas as pd
#Problem 6 List the type of schools along with the average safety score for each type.
#cps.[Elementary, Middle, or High School],NAME_OF_SCHOOL
sql_query = """ SELECT ccd.HARDSHIP_INDEX, \
                       [Elementary, Middle, or High School] AS school_type, \
                       AVG(cps.SAFETY_SCORE) AS avg_safety_score \
                FROM CHICAGO_CENSUS_DATA as ccd, \
                     CHICAGO_CRIME_DATA as cd, \
                     CHICAGO_PUBLIC_SCHOOLS as cps \
                where ccd.HARDSHIP_INDEX > 0 \
                Group by [Elementary, Middle, or High School] limit 5  
            """
pd.read_sql_query(sql_query, con)

#or
#%sql SELECT
#        cps.[Elementary, Middle, or High School] AS school_type,
#        AVG(cps.SAFETY_SCORE) AS avg_safety_score
#     FROM CHICAGO_PUBLIC_SCHOOLS AS cps
#     JOIN CHICAGO_CENSUS_DATA AS ccd
#     ON   cps.COMMUNITY_AREA_NUMBER = ccd.COMMUNITY_AREA_NUMBER \
#     НЕ НУЖНО ДЛЯ AVG where  ccd.HARDSHIP_INDEX > 0 \
#     GROUP BY cps.[Elementary, Middle, or High School];
```

```python
# сортировка по убыванию DESC
# по возрастанию SELECT * FROM имя_таблицы ORDER BY имя_столбца ASC
sql_query = """ SELECT ccd.PERCENT_HOUSEHOLDS_BELOW_POVERTY, ccd.COMMUNITY_AREA_NAME \
                FROM CHICAGO_CENSUS_DATA as ccd
                ORDER By ccd.PERCENT_HOUSEHOLDS_BELOW_POVERTY    DESC  \
                limit 5  
            """
pd.read_sql_query(sql_query, con)
```

```python
#Which community area number has the most criminal incidents (most crime-prone)?
sql_query = """ SELECT COMMUNITY_AREA_NUMBER
                FROM CHICAGO_CRIME_DATA
                GROUP BY COMMUNITY_AREA_NUMBER
                ORDER BY COUNT(*) DESC
                LIMIT 1; 
            """
pd.read_sql_query(sql_query, con)
#%sql SELECT COMMUNITY_AREA_NUMBER
#FROM CHICAGO_CRIME_DATA
#ROUP BY COMMUNITY_AREA_NUMBER
#HAVING COUNT(*) = (
#    SELECT MAX(cnt)
#    FROM (
#        SELECT COUNT(*) AS cnt
#        FROM CHICAGO_CRIME_DATA
#        GROUP BY COMMUNITY_AREA_NUMBER
#    )
#);
```

```python
#Problem 9 Use a sub-query to find the name of the community area with highest hardship index
sql_query = """ SELECT COMMUNITY_AREA_NUMBER,COMMUNITY_AREA_NAME,HARDSHIP_INDEX \
                FROM CHICAGO_CENSUS_DATA 
                where HARDSHIP_INDEX = (select MAX(HARDSHIP_INDEX) from CHICAGO_CENSUS_DATA)
            """
pd.read_sql_query(sql_query, con)
```

```python
#Problem 10 Use a sub-query to determine the Community Area Name with most number of crimes?
sql_query = """ SELECT distinct id, COMMUNITY_AREA_NAME,mAX(count_crime_COMMUNITY)
                FROM(
                SELECT ccd.ID, d.COMMUNITY_AREA_NAME,count (ccd.ID) as count_crime_COMMUNITY  \
                FROM CHICAGO_CRIME_DATA as ccd join CHICAGO_CENSUS_DATA as d \
                on   ccd.COMMUNITY_AREA_NUMBER=d.COMMUNITY_AREA_NUMBER\
                group by ccd.COMMUNITY_AREA_NUMBER \
                order by count_crime_COMMUNITY desc \
                ) 
                
            """
pd.read_sql_query(sql_query, con)
# or via LIMIT 1 without mAX(count_crime_COMMUNITY)
#sql_query = """ SELECT distinct id, COMMUNITY_AREA_NAME
#                FROM(
#                SELECT ccd.ID, d.COMMUNITY_AREA_NAME,count (ccd.ID) as count_crime_COMMUNITY  \
#                FROM CHICAGO_CRIME_DATA as ccd join CHICAGO_CENSUS_DATA as d \
#                on   ccd.COMMUNITY_AREA_NUMBER=d.COMMUNITY_AREA_NUMBER\
#                group by ccd.COMMUNITY_AREA_NUMBER \
#                order by count_crime_COMMUNITY desc \
#                LIMIT 1
#                ) 
                
#            """
```
