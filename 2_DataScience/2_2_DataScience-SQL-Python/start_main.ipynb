# SQL and Python Laboratory Menu

- **Completed by:** R. Lisovenko
- **Study period:** January–March 2026

Open a laboratory using the links below, then run its **Notebook setup** cell first.
Follow the sections in order; alternative versions are listed separately.
This menu opens notebooks without executing their exercises.

Some notebooks contain SQL or terminal examples intended for the corresponding
database console or terminal, rather than the Python kernel. Db2 laboratories
require a configured external connection. Setup scripts may drop or clear tables.

## Course folders

- `notebooks/` — laboratory notebooks
- `sql/` — SQL scripts
- `csv/` — initial data
- `DB/` — local databases
- `docs/` — documents
